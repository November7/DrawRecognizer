## Konwersja modelu Tensorflow od Teachable Machinedo formatu ONNX

Wyeksportuj model z Teachable Machine (Tensorflow savedmodel), umieść rozpakowany folder w katalogu 'tensorflow' i uruchom poniższy kod. 

Powinny być: folder z modelem i plik txt z nazwami klas (labels.txt).

```python
import shutil

teachable_machine_model_name = "./tensorflow/tm"

!python -m tf2onnx.convert --saved-model {teachable_machine_model_name} --output {teachable_machine_model_name}.onnx
```

## Dodanie etykiet klas do pliku ONNX (opcjonalne)

```python
import onnx
import shutil
import os

labels = []

with open(teachable_machine_model_name + ".txt", "r") as f:
    for line in f:
        index, name = line.strip().split()
        labels.append(name)

model = onnx.load(teachable_machine_model_name + ".onnx")

meta = model.metadata_props.add()
meta.key = "labels"
meta.value = ",".join(labels)

onnx.save(model, teachable_machine_model_name + ".onnx")

shutil.rmtree(teachable_machine_model_name)
os.remove(teachable_machine_model_name + ".txt")
```
